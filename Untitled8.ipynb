```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, adjusted_rand_score
```

```python
df = pd.read_csv("dataset.csv")
X = df[["Annual Income (k$)","Spending Score (1-100)"]].values
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled)
```

```python
class KMeansScratch:
    def __init__(self, n_clusters=5, max_iter=300,
                 tol=1e-4, random_state=42):

        self.n_clusters = n_clusters
        self.max_iter = max_iter
        self.tol = tol
        self.random_state = random_state


    # K-Means++ initialization
    def initialize_centroids(self, X, rng):

        centroids = np.empty(
            (self.n_clusters, X.shape[1])
        )

        # First centroid
        centroids[0] = X[rng.integers(len(X))]

        distances = np.sum(
            (X - centroids[0]) ** 2,
            axis=1
        )

        # Select remaining centroids
        for i in range(1, self.n_clusters):

            probabilities = distances / distances.sum()

            index = rng.choice(
                len(X),
                p=probabilities
            )

            centroids[i] = X[index]

            new_distances = np.sum(
                (X - centroids[i]) ** 2,
                axis=1
            )

            distances = np.minimum(
                distances,
                new_distances
            )

        return centroids


    # Training
    def fit(self, X):

        X = np.asarray(X, dtype=float)

        rng = np.random.default_rng(
            self.random_state
        )

        self.cluster_centers_ = (
            self.initialize_centroids(X, rng)
        )

        for iteration in range(self.max_iter):

            # Calculate distances
            distances = np.sum(
                (
                    X[:, np.newaxis, :]
                    - self.cluster_centers_[np.newaxis, :, :]
                ) ** 2,
                axis=2
            )

            # Assign each point to nearest centroid
            labels = np.argmin(distances, axis=1)

            # Calculate new centroids
            new_centroids = np.array([
                X[labels == j].mean(axis=0)
                if np.any(labels == j)
                else X[rng.integers(len(X))]
                for j in range(self.n_clusters)
            ])

            # Check convergence
            centroid_shift = np.max(
                np.linalg.norm(
                    new_centroids - self.cluster_centers_,
                    axis=1
                )
            )

            self.cluster_centers_ = new_centroids

            if centroid_shift <= self.tol:
                break

        self.n_iter_ = iteration + 1

        # Final cluster assignment
        distances = np.sum(
            (
                X[:, np.newaxis, :]
                - self.cluster_centers_[np.newaxis, :, :]
            ) ** 2,
            axis=2
        )

        self.labels_ = np.argmin(
            distances,
            axis=1
        )

        # Inertia
        self.inertia_ = np.sum(
            (
                X -
                self.cluster_centers_[self.labels_]
            ) ** 2
        )

        return self

```

```python
print("K\tInertia\t\tSilhouette")

for k in range(2, 11):

    model = KMeansScratch(
        n_clusters=k,
        random_state=42
    )

    model.fit(X_scaled)

    score = silhouette_score(
        X_scaled,
        model.labels_
    )

    print(
        f"{k}\t{model.inertia_:.2f}\t\t{score:.3f}"
    )
```

```python
k = 5
scratch_model = KMeansScratch(
    n_clusters=k,
    random_state=42
)
scratch_model.fit(X_scaled)
```

```python
sklearn_model = KMeans(
    n_clusters=k,
    init="k-means++",
    n_init=10,
    max_iter=300,
    tol=1e-4,
    random_state=42
)

sklearn_model.fit(X_scaled)
```

```python
scratch_silhouette = silhouette_score(
    X_scaled,
    scratch_model.labels_
)

sklearn_silhouette = silhouette_score(
    X_scaled,
    sklearn_model.labels_
)

ari = adjusted_rand_score(
    scratch_model.labels_,
    sklearn_model.labels_
)

print("\n--- Performance Comparison ---")

print(
    "Scratch K-Means Inertia:",
    round(scratch_model.inertia_, 4)
)

print(
    "Scikit-learn K-Means Inertia:",
    round(sklearn_model.inertia_, 4)
)

print(
    "Scratch Silhouette Score:",
    round(scratch_silhouette, 4)
)

print(
    "Scikit-learn Silhouette Score:",
    round(sklearn_silhouette, 4)
)

print(
    "Adjusted Rand Index:",
    round(ari, 4)
)
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    X[:, 0],
    X[:, 1],
    c=scratch_model.labels_,
    cmap="viridis",
    s=50
)

centers = scaler.inverse_transform(
    scratch_model.cluster_centers_
)

plt.scatter(
    centers[:, 0],
    centers[:, 1],
    marker="X",
    s=200,
    c="red",
    label="Centroids"
)

plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.title("Customer Segmentation using K-Means")
plt.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

data = pd.read_csv("dataset.csv")
scaler = StandardScaler()
data[["Annual Income (k$)","Spending Score (1-100)"]] = scaler.fit_transform(data[["Annual Income (k$)","Spending Score (1-100)"]])
X=data[["Annual Income (k$)",	"Spending Score (1-100)"]]

scores = []

for k in range(2,7):
  model = KMeans(n_clusters=k, random_state=42, n_init=10)
  labels = model.fit_predict(X)
  score = silhouette_score(X,labels)
  scores.append(score)
plt.plot(range(2,7),scores,marker='o')
plt.xlabel('Number of clusters (K)')
plt.ylabel('Average Silhouette score')
plt.title('Silhouette score for diffrent K')
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from sklearn.decomposition import PCA
```

```python
from numpy.ma import compressed
image=Image.open("imagee.jpg").convert("L")
img_array=np.array(image)
print("Original Image:",img_array.shape)
pca=PCA(n_components=0.95)
compressed_img=pca.fit_transform(img_array)
reconstructed_img=pca.inverse_transform(compressed_img)
print("Reduced Image:",reconstructed_img)
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img_array,cmap='gray')
plt.title("Original Image")
plt.subplot(1,2,2)
plt.imshow(reconstructed_img,cmap='gray')
plt.title("Compressed Image")
plt.show()

print("Original Data size",img_array.size)
print("Compressed data size",compressed_img.size)
print("Number of principal components",pca.n_components_)
print("Explained variance ratio",pca.explained_variance_ratio_.sum())
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_openml
from sklearn.decomposition import PCA
```

```python
mnist=fetch_openml('mnist_784',version=1,as_frame=False)
X=mnist.data
X=X[:5000]

print("Original dataset shape:",X.shape)

X=X/255.0

pca=PCA(n_components=2)
X_reduced=pca.fit_transform(X)

print("Reduced dataset shape:",X_reduced.shape)
X_reconstructed=pca.inverse_transform(X_reduced)
print("Reconstructed dataset shape:",X_reconstructed.shape)
```

```python
plt.figure(figsize=(12, 5))

for i in range(10):
    plt.subplot(2, 10, i + 1)
    plt.imshow(X[i].reshape(28, 28), cmap='gray')
    plt.title("Original")
    plt.axis('off')

    plt.subplot(2, 10, i + 11)
    plt.imshow(X_reconstructed[i].reshape(28, 28), cmap='gray',
               vmin=0, vmax=1)
    plt.title("Reconstructed")
    plt.axis('off')

plt.tight_layout()
plt.show()

```
